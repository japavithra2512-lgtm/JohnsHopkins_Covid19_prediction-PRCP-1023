### Importing libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load the Data
confirmed = pd.read_csv('time_series_covid19_confirmed_global.csv')
deaths = pd.read_csv('time_series_covid19_deaths_global.csv')
recovered = pd.read_csv('time_series_covid19_recovered_global.csv')
```

```python
#Convert Wide to Long Format
confirmed_long = confirmed.melt(
    id_vars=["Province/State", "Country/Region", "Lat", "Long"],
    var_name="Date",
    value_name="Confirmed"
)

deaths_long = deaths.melt(
    id_vars=["Province/State", "Country/Region", "Lat", "Long"],
    var_name="Date",
    value_name="Deaths"
)

recovered_long = recovered.melt(
    id_vars=["Province/State", "Country/Region", "Lat", "Long"],
    var_name="Date",
    value_name="Recovered"
)
```

```python
#Merging the Datasets
data = confirmed_long.merge(deaths_long,
                            on=["Province/State", "Country/Region", "Lat", "Long", "Date"])

data = data.merge(recovered_long,
                  on=["Province/State", "Country/Region", "Lat", "Long", "Date"],
                  how="left")   # recovered may have missing values
```

```python
data
```

```python
#Fix Date Type
data['Date'] = pd.to_datetime(data['Date'])
data
```

```python
# Province/State -> normal to be NaN (do not fill)
# Recovered -> fill with forward-fill
data["Recovered"] = data["Recovered"].fillna(method="ffill")

# If still NaN at beginning, fill with 0
data["Recovered"] = data["Recovered"].fillna(0)
data
```

```python
# Identify column types
numerical_cols = data.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = data.select_dtypes(include=['object']).columns.tolist()
datetime_cols = data.select_dtypes(include=['datetime64']).columns.tolist()

print("Numerical:", numerical_cols)
print("Categorical:", categorical_cols)
print("Datetime:", datetime_cols)
```

```python
data = data.sort_values(["Country/Region", "Date"])
data["Daily_Confirmed"] = data.groupby("Country/Region")["Confirmed"].diff().fillna(0)

target = data["Daily_Confirmed"]
features = data.drop(columns=["Daily_Confirmed"])
data
```

```python
# Categorical Univariate EDA
for col in categorical_cols:
    plt.figure(figsize=(12,5))

    sns.countplot(y=data[col], order=data[col].value_counts().head(20).index)
    plt.title(f'Count Plot of {col} (Top 20)')
    plt.show()

    print(f"\nUnique Values in {col}: {data[col].nunique()}")
    print("-"*80)
```

```python
# Categorical Univariate EDA
for col in categorical_cols:
    plt.figure(figsize=(12,5))

    sns.countplot(y=data[col], order=data[col].value_counts().head(20).index)
    plt.title(f'Count Plot of {col} (Top 20)')
    plt.show()

    print(f"\nUnique Values in {col}: {data[col].nunique()}")
    print("-"*80)
```

```python
# Correlation Heatmap
plt.figure(figsize=(10,6))
sns.heatmap(data[numerical_cols].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(10,6))
sns.heatmap(data[numerical_cols].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Categorical vs Numerical Bivariate EDA
for cat in categorical_cols:
    for num in numerical_cols:
        plt.figure(figsize=(12,5))
        sns.barplot(x=data[cat], y=data[num], estimator='mean',
                    order=data[cat].value_counts().head(10).index)
        plt.title(f'Mean {num} by {cat} (Top 10)')
        plt.xticks(rotation=45)
        plt.show()

        plt.figure(figsize=(12,5))
        sns.boxplot(x=data[cat], y=data[num],
                    order=data[cat].value_counts().head(10).index)
        plt.title(f'{num} distribution by {cat} (Top 10)')
        plt.xticks(rotation=45)
        plt.show()
```
